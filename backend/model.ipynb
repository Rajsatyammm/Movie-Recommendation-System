# EDA (Exploratory Data Analysis)

```python
# Drop null title value
df = df.dropna(subset=['title'])
```

```python
# Fill null value of overview with empty space
df['overview'] = df['overview'].fillna("")
df['tagline'] = df['tagline'].fillna("")
```

```python
import ast
from typing import List

# Function to extract genres as a string
def extract_genres(value: List) -> str:
    value = ast.literal_eval(value)
    return " ".join(dic['name'] for dic in value)

df['genres'] = df['genres'].apply(extract_genres)
```

```python
df.head()
```

```python
df['tags'] = df['overview'] + " " + df['genres'] + " " + df['tagline']
```

```python
df.head()
```

### Importing NLTK Library

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmetizer = WordNetLemmatizer()
```

```python
import re

def preprocess_text(text):
    text = str(text).lower()
    # remove punctuation
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()
    # remove stop words
    words = [word for word in words if word not in stop_words]

    words = [lemmetizer.lemmatize(word) for word in words]
    return " ".join(words)

df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df.head()
```

```python
df.shape
```

```python
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index=df['title'])
indices
```

### Apply Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1, 2), stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
    if title not in indices:
        return ['movie not found']
    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]
    return df['title'].iloc[similar_idx]
```

```python
recommend("Toy Story", 3)
```

### Exporting the necessary files

```python
import pickle

pickle.dump(tfidf_matrix, open('tfidf_matrix.pkl', 'wb'))
pickle.dump(indices, open('indices.pkl', 'wb'))
df.to_pickle('df.pkl')
pickle.dump(tfidf, open('tfidf.pkl', 'wb'))
```
